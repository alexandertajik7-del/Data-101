```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("solomonameh/spotify-music-dataset")

print("Path to dataset files:", path)
```

```python
#### Identify Missing Data ####

# Importing necessary libraries
import pandas as pd

# Load the dataset
data = pd.read_csv('high_popularity_spotify_data2.csv')

# Count the number of missing values in energy
missing_energy = data['energy'].isnull().sum()

# Count the total number of rows in the dataset
total_rows = data.shape[0]

# Calculate the percentage of missing energy values
percentage_missing_energy = (missing_energy / total_rows) * 100

# Display the results
print(f'Number of missing energy values: {missing_energy}')
print(f'Total number of rows: {total_rows}')
print(f'Percentage of missing energy values: {percentage_missing_energy:f}%')
```

```python
# Count the number of missing values in tempo
missing_tempo = data['tempo'].isnull().sum()

# Count the total number of rows in the dataset
total_rows = data.shape[0]

# Calculate the percentage of missing tempo values
percentage_missing_tempo = (missing_tempo / total_rows) * 100

# Display the results
print(f'Number of missing energy values: {missing_tempo}')
print(f'Total number of rows: {total_rows}')
print(f'Percentage of missing tempo values: {percentage_missing_tempo:f}%')
```

```python
# Count the number of missing values in danceability
missing_danceability = data['danceability'].isnull().sum()

# Count the total number of rows in the dataset
total_rows = data.shape[0]

# Calculate the percentage of missing danceability values
percentage_missing_danceability = (missing_danceability / total_rows) * 100

# Display the results
print(f'Number of missing danceability values: {missing_danceability}')
print(f'Total number of rows: {total_rows}')
print(f'Percentage of missing danceability values: {percentage_missing_danceability:f}%')
```
